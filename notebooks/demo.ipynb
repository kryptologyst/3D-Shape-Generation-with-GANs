# 3D Shape Generation Demo

This notebook demonstrates how to use the 3D Shape Generation framework to train and evaluate GAN models for generating 3D point clouds.


```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Import our modules
from src.models.gan_models import create_model, initialize_weights
from src.data.datasets import SyntheticDataset
from src.train.trainer import Trainer
from src.utils.visualization import PointCloudVisualizer
```

## 1. Create Models

Let's create a point cloud GAN with a generator and discriminator.


```python
# Model parameters
latent_dim = 128
num_points = 1024
device = 'cuda' if torch.cuda.is_available() else 'cpu'

print(f"Using device: {device}")

# Create models
generator, discriminator = create_model(
    model_type="point_cloud",
    latent_dim=latent_dim,
    num_points=num_points
)

# Initialize weights
initialize_weights(generator)
initialize_weights(discriminator)

# Move to device
generator = generator.to(device)
discriminator = discriminator.to(device)

print(f"Generator parameters: {sum(p.numel() for p in generator.parameters()):,}")
print(f"Discriminator parameters: {sum(p.numel() for p in discriminator.parameters()):,}")
```
